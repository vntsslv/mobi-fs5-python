**[Task 2.2.]()** Implement the two functions `sobel_h` and `sobel_v` for horizontal and vertical $3 \times 3$ Sobel filtering, respectively:

```python
def sobel_h(img):
    gx = np.zeros(img.shape)
    gx[1:-1, 1:-1] = (
        img[:-2, 2:] + 2*img[1:-1, 2:] + img[2:, 2:]
        - img[:-2, :-2] - 2*img[1:-1, :-2] - img[2:, :-2]
    )
    return gx
    
def sobel_v(img):
    gy = np.zeros(img.shape)
    gy[1:-1, 1:-1] = (
        img[2:, :-2] + 2*img[2:, 1:-1] + img[2:, 2:]
        - img[:-2, :-2] - 2*img[:-2, 1:-1] - img[:-2, 2:]
    )
    return gy
```

**[Task 2.3.]()** Test your implementations from above:

```python
gx = sobel_h(img)
gy = sobel_v(img)

plt.figure(figsize=(15, 6))

plt.subplot(1, 2, 1)
plt.imshow(gx, cmap='gray')
plt.colorbar()
plt.title('Horizontal derivative')

plt.subplot(1, 2, 2)
plt.imshow(gy, cmap='gray')
plt.colorbar()
plt.title('Vertical derivative')

plt.show()
```
